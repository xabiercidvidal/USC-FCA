## 1. Introduction to functional programming

**Functional Programming**, FP, is a programming paradigm that defines a program as a flow of functions evaluated (in general) in a series. See [Functional Programming HOWTO](https://docs.python.org/3/howto/functional.html)

 - Each function takes an input and produces an output, it has no side effects: it always produces the same output for the same input (a *pure* function).

 - Data has no evolving state. It does not change along the program. Data just flows between functions.

 - FP code is modular, reusable, and safe.

 - Functions are small pieces of code, they can be re-arranged to create a new program and they are easy to check.

 - Functions must be tested. Create tests for each function. In fact, the tests *define* the function (see the notebook about testing).

In FP the code is:

**High-level**: You focus on describing the desired outcome rather than detailing the exact steps to achieve it. These concise statements are powerful and effective.

**Transparent**: The behavior of a pure function is determined solely by its inputs and outputs, without involving intermediary states. This eliminates side effects, making debugging simpler.

**Parallelizable**: Functions without side effects are easier to run in parallel, allowing for more efficient execution.

Python nicely supports functional programming, mostly via **comprehensions**, **iterators**, **generators**, a large functionality for **functions**, and the standard modules **functools** and **itertools**.

## 2. Some useful utilities

###  2.1 *zip*

The *zip()* builtin function pairs the items of several iterables: it produces tuples with the first items of each, then the second ones, etc.

This is an example of how to use *zip()*:

```python
xs  = list(range(0, 3))
ys  = list(range(3, 6))
xys = list(zip(xs, ys))
print(xs)
print(ys)
print(xys)

for xi, yi in zip(xs, ys):
    print(' xi, yi ', xi, yi)
```

**Careful**: if the iterables have different lengths, *zip()* silently stops at the shortest one. Since Python 3.10 you can ask it to check the lengths with `strict=True`.

The expression `zip(*pairs)` does the inverse operation: it "unzips" a list of pairs (it is a transposition).

```python
print(list(zip([1, 2, 3], ['a', 'b'])))          # the 3 is lost without warning!

try:
    list(zip([1, 2, 3], ['a', 'b'], strict=True))
except ValueError as error:
    print('ValueError:', error)

points = [(0., 1.), (2., 3.), (4., 5.)]
xs, ys = zip(*points)
print('xs =', xs, ' ys =', ys)
```

### 2.2 *enumerate*

Another important iterator is generated by the *enumerate()* builtin function. *enumerate()* takes an iterable and returns an iterator of pairs: the index position of the item and the item itself.

This is particularly useful when inside a loop one needs access to both the index position of the item in the list and the item.

Let's see the example:

```python
xs = ['a', 'b', 'c', 'd']

for i, xi in enumerate(xs):
    print('item [', i, '] = ', xi)

# the counter can start at any value
for i, xi in enumerate(xs, start=1):
    print('item number', i, ':', xi)
```

### 2.3 Iterating over dictionaries

Iterating over a dictionary gives its keys; *.values()* gives the values and *.items()* the (key, value) pairs:

```python
masses = {'electron': 0.511, 'muon': 105.66, 'tau': 1776.9}   # MeV

for name, mass in masses.items():
    print(f'{name:10s} {mass:8.2f} MeV')
```

## 3. List comprehensions

List comprehensions build a list by applying an expression to the items of an iterable, optionally keeping only the items that fulfil a condition (a *predicate*).

List comprehensions are key ingredients of Functional Programming in Python.

The syntax:

  * `[expression(item) for item in iterable]`

  * `[expression(item) for item in iterable if condition(item)]`

### 3.1 Comprehensions, *map* and *filter*

List comprehensions are an elegant (and usually more readable) version of the *map* and *filter* builtin functions.

```python
xs = list(range(6))
print('list ', xs)

# compute the square of the items on the list
x2as = list(map(lambda xi: xi * xi, xs))
print('x2 ', x2as)

# the same!
x2bs = [xi * xi for xi in xs]
print('x2 ', x2bs)
```

```python
print('initial list ', xs)

# filter the even numbers on the list
xevens = list(filter(lambda xi: xi % 2 == 0, xs))
print('evens ', xevens)

# the same!
xevens = [xi for xi in xs if xi % 2 == 0]
print('evens ', xevens)
```

List comprehensions can loop over several iterables. The general syntax is:

`[expression(item1, item2, ...) for item1 in iterable1 for item2 in iterable2 ... ]`

They can be combined with conditions:

`[expression(item1, item2, ...) for item1 in iterable1 if condition(item1) for item2 in iterable2 if condition(item2) ...]`

The `for` clauses are read in the same order as nested *for* loops: the first one is the outer loop.

```python
xs = list(range(3))
ys = list(range(3,6))
print('xs ', xs)
print('ys ', ys)

xys = [(xi, yi) for xi in xs for yi in ys]
print('xys ', xys)

xyevens = [(xi, yi) for xi in xs if xi%2 == 0 for yi in ys if yi % 2 != 0]
print('xy-even ', xyevens)

xyevens = []
for xi in xs:
    if (xi % 2 == 0):
        for yi in ys:
            if (yi % 2 != 0):
                xyevens.append( (xi, yi))
print('the same :', xyevens)
```

Do not confuse a comprehension with two *for* clauses (a flat list) with a comprehension *inside* another one (a list of lists, e.g. a matrix):

```python
flat   = [i * j for i in range(3) for j in range(3)]
matrix = [[i * j for j in range(3)] for i in range(3)]
print(flat)
print(matrix)
```

A comprehension can also loop over several iterables *at the same time*, using the *zip()* builtin function.

The general expression is:

`[expression(item1, item2) for item1, item2 in zip(iterable1, iterable2) if condition(item1, item2)]`

```python
xs = list(range(3))
ys = list(range(3, 6))
print('xs ', xs)
print('ys ', ys)

xys = [yi - xi for xi, yi in zip(xs, ys) if xi%2 == 0]
print('xys ', xys)
```

Another possibility is to use *map()*, which accepts several iterables, with a *lambda* function.

```python
xs = list(range(3))
ys = list(range(3, 6))
print('xs =', xs)
print('ys =', ys)

xys = list(map(lambda xi, yi: yi + xi, xs, ys))
print('xys =', xys)
```

Some advice on style:

  * Use a comprehension to *build a collection*, not for its side effects (`[print(x) for x in xs]` builds a useless list of *None*: use a *for* loop).
  * If a comprehension needs more than two *for* clauses or complicated conditions, a normal loop (or a helper function) is more readable.
  * For numerical work on large arrays of numbers, numpy vectorised operations are much faster than any comprehension (see the numpy notebook).

### 3.2 Dictionary and set comprehensions, generator expressions

The same syntax works with curly braces, to build dictionaries and sets:

```python
elements = ['H', 'He', 'Li', 'Be']
Z        = [1, 2, 3, 4]

atomic_number = {el: z for el, z in zip(elements, Z)}
print(atomic_number)

charges = [1, -1, 1, 0, -1, 2]
print('different charges:', {q for q in charges})
```

With parentheses, instead, we obtain a **generator expression**: the values are *not* stored, they are produced one by one when they are needed (see sections 5 and 6). When the generator expression is the only argument of a function, the extra parentheses can be dropped:

```python
import sys

n        = 1_000_000
as_list  = [i * i for i in range(n)]
as_gen   = (i * i for i in range(n))

print('size of the list      :', sys.getsizeof(as_list), 'bytes')
print('size of the generator :', sys.getsizeof(as_gen), 'bytes')
print('sum of squares        :', sum(i * i for i in range(n)))
```

### 3.3 Sorting a list

The *sorted()* builtin function returns a new, sorted list.

```python
xs = [8, 10, 9, 5, 3, 1]
print('original ', xs)

oxs = sorted(xs)
print ('ordered ', oxs)
print ('first list ', xs)
```

*sorted()* accepts two optional arguments. We can reverse the ordering by setting *reverse=True*.

The argument *key* is a function that is applied to each item to get the value that will be used to order the items.

Let's see some examples:

```python
xs = list(range(5))
print('original list', xs)

# reverse ordering
ixs = sorted(xs, reverse = True)
print('inverse ordering ', ixs)
```

```python
# using a function to get the value in which the ordering will be performed
def dis2(x, x0 = 2.):
    """ return the absolute distance of x with respect x0
    """
    return abs(x-x0)

x2s = sorted(xs, key = dis2)
print ('initial list ', xs)
print ('ordered respect to 2', x2s)
```

The key function is very often a *lambda*. If the key returns a tuple, the items are sorted by the first element, then by the second one, etc. Sorting in Python is *stable*: items with equal keys keep their original order.

*min()* and *max()* also accept a *key* argument.

```python
# (name, mass [MeV], momentum [GeV])
particles = [('pi+', 139.6, 2.3), ('K-', 493.7, 1.1), ('p', 938.3, 5.4), ('pi-', 139.6, 0.8)]

print('by momentum     :', sorted(particles, key=lambda p: p[2], reverse=True))
print('by mass, then p :', sorted(particles, key=lambda p: (p[1], p[2])))
print('heaviest        :', max(particles, key=lambda p: p[1]))
```

NB One can also sort a list *in place*, without creating a new one, with the *.sort()* method. Note that it returns *None*, so `xs = xs.sort()` is a bug!

```python
print("before", ixs)
result = ixs.sort()
print("after ", ixs)
print("sort() returned", result)
```

### 3.4 Reductions

Some builtin functions reduce an iterable to a single value: the sum of all items, *sum()*, the maximum, *max()*, or the minimum, *min()*.

In addition, for booleans, *any()* returns *True* if at least one item is *True*, and *all()* returns *True* if all items are *True*. Both stop as soon as the answer is known (*short-circuit*), which is very efficient combined with a generator expression.

```python
xs  = list(range(1, 10))
xsi = sorted(xs, reverse = True)
print('xsi = ', xsi)

print('sum(xsi) = ', sum(xsi))
print('max(xsi) = ', max(xsi))
print('min(xsi) = ', min(xsi))
```

```python
bs = [True, True, False]
print ('bs = ', bs)
print ('any(bs) = ', any(bs))
print ('all(bs) = ', all(bs))

print('any item larger than 5? ', any(x > 5 for x in xs))
print('all items positive?     ', all(x > 0 for x in xs))
```

The general tool is *functools.reduce(f, iterable)*, which applies a function of two arguments cumulatively: $f(f(f(x_0, x_1), x_2), x_3)\dots$ For the product there is also *math.prod()*.

```python
import math
from operator  import mul
from functools import reduce

xs = list(range(1, 5))
print('xs = ', xs)

## multiply
p = reduce(mul, xs)
print('product of all items in list =', p, ' math.prod:', math.prod(xs))

## sum
print("check=",reduce(lambda x,y: x+y,xs)-sum(xs))
```

Remember the function *create_polynomial*; we can re-write it now with a generator expression:

```python
from collections.abc import Callable

def create_polynomial(cas: tuple[float, ...]) -> Callable[[float], float]:

    def pol(x: float) -> float:
        return sum(ai * x**i for i, ai in enumerate(cas))

    return pol
```

```python
cas = (1., +1., 1.)
pol = create_polynomial(cas)
x = 2
print(pol(x))
```

We can use list comprehensions to write more elegant code:

```python
def nfactorial(n):
    """ returns n! = n*(n-1)*...*1
    """
    if (n <= 1 ):
        return 1
    return n * nfactorial(n-1)

ns = [nfactorial(ni) for ni in range(10)]
print(ns)
```

## 4. The *itertools* module

The [itertools](https://docs.python.org/3/library/itertools.html) standard module provides efficient building blocks to combine iterables. Some of the most useful ones in physics:

  * *product*: cartesian product, replaces nested loops (grids of parameters).
  * *combinations*: all unique pairs (or triplets...) of items, e.g. pairs of particles for a two-body interaction.
  * *permutations*: all the orderings.
  * *accumulate*: running sums (or running products, maxima...).
  * *chain*: concatenates iterables.
  * *count*, *islice*: infinite counters, and taking a slice of any iterator (see section 6).

```python
import itertools

# product: all the points of a grid, no nested loops
print('grid         :', list(itertools.product([0, 1], [0, 1, 2])))

# combinations: unique pairs, i < j
charges = [('q1', 1.), ('q2', -1.), ('q3', 2.)]
print('pairs        :', [(a[0], b[0]) for a, b in itertools.combinations(charges, 2)])

# permutations
print('permutations :', list(itertools.permutations('abc')))

# accumulate: running sum, e.g. the positions of a walker given its steps
steps = [1, -1, -1, 1, 1, 1]
print('positions    :', list(itertools.accumulate(steps)))

# chain
print('chain        :', list(itertools.chain([1, 2], (3, 4), 'ab')))
```

## 5. Iterators

In many cases you want to run a function over a set of items, a stream. Iterators are objects that represent that stream.

An iterator returns the next object in the stream when the *next()* builtin function is applied to it. When the end of the stream is reached, the iterator raises a *StopIteration* exception.

An object is *iterable* if it can provide an iterator.

Lists, tuples, dictionaries and strings are iterables. The way to get the iterator is via the *iter()* builtin function.

When dealing with *for* statements, we are in fact dealing with iterators. Here is a 'low level' code that shows how the list iterator works. When you are using the *for* statement, Python does all this work for you!

```python
x   = list(range(3))
itx = iter(x)
print('iter(x) is of type ', type(itx))
print('list is ', x)

go = True
while go:
    try:
        xi = next(itx)
        print('item ', xi)
    except StopIteration:
        go = False
        print('end of iteration reached!')

for xi in x:
    print('item ', xi)
```

Something similar happens with strings:

```python
x = 'Hello!'
itx = iter(x)
print('string is ', x)

go = True
while go:
    try:
        xi = next(itx)
        print('letter ', xi)
    except StopIteration:
        go = False
        print('end of iteration reached!')

for xi in x:
    print('letter ', xi)
```

### 5.1 Lists from iterators

We can recover a list (or tuple) from an iterator. *list(iterator)* will produce the list with the remaining items of the iterator.

```python
ns = range(4)
print('original range ', ns)

# getting the iterator
it = iter(ns)
print('first item in the iterator ', next(it) )

# getting the list of the rest of the iterator
xns = list(it)
print('a list with the other items ', xns)
```

### 5.2 Iterators are consumed

An iterator can be traversed **only once**. *map*, *filter*, *zip*, *enumerate*, generator expressions and generators all return iterators: the second time you loop over them they are empty. A *range*, instead, is a lazy *sequence* (it computes its items on demand, but it can be traversed many times, like a list).

```python
squares = map(lambda x: x * x, range(4))
print('first time :', list(squares))
print('second time:', list(squares))     # empty!

r = range(4)
print('a range can be reused:', list(r), list(r))
```

## 6. Generators

Generators are special functions that return an iterator. Instead of *return*, they use *yield*: each time *next()* is applied, the function runs until the next *yield*, returns that value, and is *paused* there, keeping all its local variables. When the function ends (or reaches a *return*), a *StopIteration* exception is raised.

This is an example of a generator of even numbers:

```python
def generator_even(n):
    """ yields the even numbers in [0,..., n-1]
    """
    for ni in range(0, n, 2):
        yield ni
```

```python
# create the generator of even numbers from 0 to 10
it = generator_even(10)
print('type ', it)

# get the first item in the iterator
print('item! = ', next(it))

# loop in the rest of item in the iterator
for itv in it:
    print('item =', itv)
```

Note that the numbers are produced one by one: they are never stored in a list. This makes generators ideal for **infinite streams**, e.g. the successive states of a simulation. We can take a finite number of items with *itertools.islice*:

```python
import random

def random_walk(x0=0, p=0.5):
    """ infinite 1D random walk: yields the successive positions of the walker
    """
    x = x0
    while True:
        yield x
        x += 1 if random.random() < p else -1

random.seed(1)
print(list(itertools.islice(random_walk(), 15)))
```

**Generator pipelines.** Generators (and generator expressions) can be chained. Each stage processes one item at a time, so we can analyse files much larger than the memory. Here we simulate a data file with *io.StringIO*; with a real file, just replace it by `open('data.txt')`, which is also an iterator over lines:

```python
import io

data = io.StringIO("""# t   E
0.0  1.2
0.1 -0.5
0.2  3.4
# detector off
0.3  2.2
""")

lines    = (line for line in data if not line.startswith('#'))     # skip comments
values   = (tuple(map(float, line.split())) for line in lines)     # parse
positive = (v for v in values if v[1] > 0)                         # select

for t, E in positive:        # nothing is read until here!
    print(f't = {t:.1f}  E = {E:.1f}')
```

### 6.1 Advanced (optional): sending values to a generator

A generator can also *receive* values: `value = yield x` pauses the generator, and `it.send(v)` resumes it, making the *yield* expression evaluate to *v* (with a plain *next()* it evaluates to *None*). *close()* ends the generator. This mechanism (coroutines) is the basis of *asyncio* (see the concurrency notebook).

In this example, *send(v)* makes the generator jump to the multiple of *step* just below *v*:

```python
def stepper(n, step=2):
    """ yields 0, step, 2*step, ... below n.
    send(v) jumps to the multiple of step just below v
    """
    ni = 0
    while ni < n:
        val = yield ni
        if val is not None and val < n:
            ni = step * (val // step)
        else:
            ni = ni + step
```

```python
print('all values ', tuple(stepper(10, 3)))

it = stepper(10, 3)
print('first item is ', next(it))
print('jump to 7, item now is ', it.send(7))
print('next item is ', next(it))

# close the iteration
it.close()
```

## Summary

We have seen that Python provides powerful tools for Functional Programming:

  1. *zip* and *enumerate* combine iterables; beware that *zip* stops at the shortest one.
  2. List comprehensions `[expression for item in iterable if condition]` replace *map* and *filter*; the same syntax builds dictionaries, sets and generator expressions.
  3. *sorted*, *min* and *max* accept a *key* function; *.sort()* sorts in place and returns *None*.
  4. *sum*, *any*, *all*, *math.prod* and *functools.reduce* reduce an iterable to a single value.
  5. *itertools* provides products, combinations, running sums, slices of iterators...
  6. Lists, tuples, dictionaries and strings are iterable objects; iterators are traversed only once.
  7. Generators (*yield*) produce values lazily: they allow infinite streams and memory-efficient pipelines.
