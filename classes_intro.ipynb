## 1. Introduction to Classes

### 1.1 What is a class, what is an object?

What is a car? Which are its elements, which are its actions? These two characteristics, elements and actions, define a 'generic' car. In the real world we have concrete cars: polos, minis and beetles.

The 'virtual' generic definition of 'car' is its **class** (or **type**), while the concrete cars are **objects** (or **instances**). My own car is an object.

<img src="./img/classes_car.png" width="400" height="400">

A **class** is a piece of code that defines a type: its **attributes** (the data) and its **methods** (the actions, functions that act on the data).

Let's consider a complex number. It is a type: it has a real and an imaginary part (its attributes), and it accepts several operations: addition, subtraction, product, modulus, etc. (its methods).

An **object** is a concrete instance of a class. For example, in *x = 1 + 2j*, *x* is a variable associated to an object whose type, or class, is *complex*.

The method that creates an object of a class is called the **constructor**.

### 1.2 Everything is an object

In Python everything is an object: numbers, strings, lists, functions, modules... *type()* tells us the class of an object:

```python
for obj in [3, 3.5, 'text', [1, 2], (1, 2), {'a': 1}, len, math.sqrt]:
    print(f'{obj!r:25} {type(obj)}')
```

And all of them have attributes and methods, which we access with a dot: *object.attribute* and *object.method(arguments)*.

We have already used some: the methods of lists and strings, and the attributes of functions in the notebook about decorators (*\_\_name\_\_*, *\_\_doc\_\_*).

```python
print('text'.upper())
print((3.5).is_integer())
print(len.__name__)
print(len.__doc__)
```

To check the type of an object, *isinstance(object, class)* is usually better than *type(object) == class*. It accepts a tuple of classes, and also accepts derived classes (we will see what that means with inheritance):

```python
x = 3
print(isinstance(x, (int, float)))
print(isinstance(True, int), type(True) == int)    # bool is derived from int
```

### 1.3 The complex class as example

#### Constructor and type

```python
x = 1 + 2j
type(x)
```

Notice that the *'='* operator creates an object of type *complex* and assigns it to the variable *x*.

We can also create it by calling the constructor, which has the name of the class:

```python
x = complex(1., 1.)
print(x)
print('x is of type ', type(x))
```

#### Getting help

We can get information about a class and its methods with *help()*: *help(complex)* shows everything, *help(complex.conjugate)* a single method.

```python
help(complex.conjugate)
```

*dir(object)* lists all its attributes and methods. The names that start with two underscores are special; we will see them below. The rest are:

```python
[name for name in dir(x) if not name.startswith('_')]
```

####  Accessing the attributes

We access the attributes, the data, of an object via the *'.'* operator. The attributes of *complex* are *real* and *imag*:

```python
x = 2 + 1j
print('x = ', x)
a, b = x.real, x.imag
print('x.real =', a, ', x.imag = ', b)
```

The attributes of a complex number cannot be changed: *complex* is an *immutable* type, like *int*, *float*, *str* or *tuple*. When we write our own classes, we will decide which attributes can change.

```python
try:
    x.real = 3
except AttributeError as error:
    print('AttributeError:', error)
```

#### Applying some methods

How to conjugate a complex number? Use the method *conjugate*, with the '.' syntax:

*object.method(arguments)*

```python
xc = x.conjugate()
print('x = ', x, ', x^c = ', xc)
print('xc is of type ', type(xc))
xcc = xc.conjugate()
print(xcc)
```

An alternative way is to call the function *complex.conjugate()* with the object *x* as argument.

A method is just a function defined inside the class: the class is the namespace where the function lives, like a module. *x.conjugate()* is the same as *complex.conjugate(x)*: Python passes the object as the first argument. When we write our own classes, this first argument will be called *self*.

```python
x = 1 - 1j
print(x.conjugate())
print(complex.conjugate(x))
```

#### Special methods and builtin functions

The methods whose names start and end with two underscores, like *\_\_abs\_\_*, are special: they are associated to a builtin function or to an operator. *\_\_abs\_\_* is called by the builtin function *abs()*, which computes the modulus of the complex number.

The three expressions below are equivalent, but the most elegant is the first one.

```python
x = complex(1, 1)
print(abs(x))
print(x.__abs__( ))
print(complex.__abs__(x))
```

The operators are special methods too. *x + y* calls *x.\_\_add\_\_(y)*. This allows us to use a mathematical syntax. *complex* also has *\_\_sub\_\_*, *\_\_mul\_\_*, *\_\_truediv\_\_*, *\_\_eq\_\_*...

```python
x = -1 + 1j
y = complex(1, -3)
z = x + y ## x + y
print(z)
print(type(z))
print(x.__add__(y))
print(complex.__add__(x,y))
```

The same is true for the containers: *len()*, indexing *[ ]* and the operator *in* call special methods.

```python
data = [10, 20, 30]
print(len(data),     data.__len__())
print(data[1],       data.__getitem__(1))
print(20 in data,    data.__contains__(20))
print(data == [10, 20, 30], data.__eq__([10, 20, 30]))
```

#### Serializing the object

The special methods *\_\_str\_\_* and *\_\_repr\_\_* convert the object into a string:

  * *str(x)*, used by *print()*, is meant to be readable.
  * *repr(x)*, used by the notebook to show the last value of a cell, is meant to be unambiguous: ideally, the code that creates the object.

For *complex* they coincide. For strings they do not:

```python
x = -1 + 1j
print(str(x), repr(x))

s = 'hello'
print(str(s), repr(s))
s
```

### 1.4 Same operator, different classes

Look at the following function. Does it work for integers, floats, complex numbers, lists, strings, numpy arrays?

```python
def add(a, b):
    return a + b

print(add(1, 2))
print(add(1 + 1j, 2))
print(add([1, 2], [3, 4]))
print(add('ab', 'cd'))
print(add(np.array([1, 2]), np.array([3, 4])))
```

Yes! Python does not care about the type of *a* and *b*; it only needs that *a + b* is defined, that is, that the class of *a* has a method *\_\_add\_\_* (this is *duck typing*, as we saw in the notebook about functions).

But each class decides what *+* means: lists concatenate, numpy arrays add element by element. In the next notebook we will write our own classes, with their own *\_\_add\_\_*.

A numpy array is a good example of an object with attributes (no parentheses) and methods (with parentheses):

```python
a = np.array([[1., 2., 3.], [4., 5., 6.]])
print(a.shape, a.ndim, a.dtype)      # attributes
print(a.sum(), a.mean(), a.max())     # methods
```

## Summary

  * A **class** defines a type: its attributes (data) and methods (functions acting on the data). An **object** is a concrete instance of a class, created by the constructor.

  * In Python everything is an object. *type()* and *isinstance()* tell us its class; *help()* and *dir()* what it contains.

  * *object.method(args)* is the same as *Class.method(object, args)*.

  * Special methods (*\_\_add\_\_*, *\_\_abs\_\_*, *\_\_len\_\_*, *\_\_str\_\_*...) are called by operators and builtin functions. Each class decides what they do.
